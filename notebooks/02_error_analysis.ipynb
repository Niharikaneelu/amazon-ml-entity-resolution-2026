# Error analysis
Review the highest-confidence mistakes and threshold behavior.

```python
from pathlib import Path
import pandas as pd

predictions = pd.read_csv(Path.cwd().parent / 'output' / 'predictions.csv')
predictions['error'] = predictions['match'] != predictions['label']
errors = predictions[predictions['error']].sort_values('match_probability', ascending=False)
errors.head(20)
```

```python
thresholds = [0.3, 0.5, 0.7, 0.9]
pd.DataFrame({
    'threshold': threshold,
    'predicted_matches': (predictions['match_probability'] >= threshold).sum(),
    'precision': ((predictions['match_probability'] >= threshold) & (predictions['label'] == 1)).sum() / max((predictions['match_probability'] >= threshold).sum(), 1)
} for threshold in thresholds)
```
