# Noise analysis
Inspect missingness and field-level disagreement in labeled pairs.

```python
from pathlib import Path
import pandas as pd

train_path = Path.cwd().parent / 'dataset' / 'train.csv'
pairs = pd.read_csv(train_path)
pairs.head()
```

```python
paired_fields = [column[:-5] for column in pairs if column.endswith('_left') and column[:-5] + '_right' in pairs]
summary = pd.DataFrame({
    field: {
        'missing_left': pairs[f'{field}_left'].isna().mean(),
        'missing_right': pairs[f'{field}_right'].isna().mean(),
        'exact_agreement': (pairs[f'{field}_left'].fillna('').astype(str).str.lower() == pairs[f'{field}_right'].fillna('').astype(str).str.lower()).mean(),
    } for field in paired_fields
).T
summary
```
